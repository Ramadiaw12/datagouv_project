## 🎯 01 — Contexte du projet

Dans un système d'information hospitalier, les données relatives à un même patient peuvent être réparties entre plusieurs applications : gestion des admissions, dossier patient informatisé et facturation.

Ces applications peuvent contenir des informations incomplètes, incohérentes ou dupliquées. L'absence d'une vision unifiée complique l'identification fiable des patients et la consolidation de leurs informations.

**Problématique :** comment rapprocher les enregistrements issus de plusieurs sources, détecter les incohérences et construire un référentiel patient fiable, traçable et réversible ?

## 🚀 02 — Objectifs

- **Profiler les sources** afin de comprendre leur structure, leur volumétrie et la qualité de leurs données.
- **Évaluer la Data Quality** à travers la complétude, l'unicité, les doublons et les incohérences.
- **Réaliser le matching** pour identifier les enregistrements correspondant à un même patient.
- **Construire un Golden Record** en consolidant les informations provenant des différentes applications.
- **Garantir la traçabilité** des rapprochements, des corrections et des décisions de fusion.

## 1.6 Calculer le Data Quality Score

```python
POIDS = {"completude": 25, "validite": 20, "coherence": 20, "actualite": 10, "unicite": 25}
def dqs(scores, poids=POIDS):
    """DQS = Σ(Si·Wi) / ΣWi, calculé sur les seules dimensions mesurées (score non None)."""
    mesures = {d: s for d, s in scores.items() if s is not None}
    return sum(s * poids[d] for d, s in mesures.items()) / sum(poids[d] for d in mesures)
def niveau(score):
    """Grille interne du cours."""
    return "Gold" if score >= 95 else "Silver" if score >= 85 else "À améliorer"
# Exemple : une source où l'actualité n'est pas mesurable
exemple = {"completude": 96.7, "validite": 97.5, "coherence": 97.0, "actualite": None, "unicite": 97.2}
print(round(dqs(exemple), 1), niveau(dqs(exemple)))
# DQS d'une source puis DQS global « patient » (moyenne pondérée par le nombre de lignes)
dqs_adm = dqs(scores_adm)
# à compléter avec vos scores
# dqs_global = (dqs_adm * len(adm) + dqs_dp_pat * len(dp_pat) + dqs_fac * len(fac)) / (len(adm)
# + len(dp_pat) + len(fac))
```

## 2.2 Standardisation

```python
def normaliser(texte):
    if pd.isna(texte):   # valeur manquante : chaîne vide
        return ""
    # minuscules, accents séparés des lettres
    t = unicodedata.normalize("NFD", str(texte).lower())
    t = "".join(c for c in t if unicodedata.category(c) != "Mn") # suppression des accents
    # suppression espaces et ponctuation
    return re.sub(r"[^a-z0-9]", "", t)
    
# Table de correspondance : 15 libellés de service -> 5 services (décision métier à valider)

SERVICE_MAP = {
    "Cardiologie": "CARDIO", "Service Cardiologie": "CARDIO", "CARD-01": "CARDIO",
    "Pneumologie": "PNEUMO", "Service Pneumologie": "PNEUMO", "PNEU-02": "PNEUMO",
    "Neurologie": "NEURO", "Service Neurologie": "NEURO", "NEUR-03": "NEURO",
    "Urgences": "URGENCES", "Service Urgences": "URGENCES", "URG-04": "URGENCES",
    "Pédiatrie": "PEDIATRIE", "Service Pédiatrie": "PEDIATRIE", "PED-05": "PEDIATRIE",
}
SEXE_MAP = {"M": "M", "F": "F", "1": "M", "2": "F"}    # codage unifié du sexe


# Fonction 2 : vers_pivot(...) -> tableau au format commun (« modèle pivot »)
# Les trois sources n'ont ni les mêmes colonnes ni les mêmes formats. Cette fonction convertit
# UNE source en tableau de colonnes identiques : src, rec_id, nom, prenom, dob, annee, sexe,
# service, true_id.
# Paramètres : df = la source ; src = son étiquette ; id_col, nom, prenom, dob = noms de ses
# colonnes ;
# fmt = format de sa date de naissance ; sexe = nom de la colonne sexe (None si elle n'existe
# pas).
def vers_pivot(df, src, id_col, nom, prenom, dob, fmt, sexe=None):
    # origine et identifiant de l'enregistrement
    out = pd.DataFrame({"src": src, "rec_id": df[id_col]})
    out["nom"], out["prenom"] = df[nom], df[prenom]
    # lecture selon le format de la source ; NaT si date impossible
    d = pd.to_datetime(df[dob], format=fmt, errors="coerce")
    # date réécrite au format ISO unique
    out["dob"] = d.dt.strftime("%Y-%m-%d")
    # année lue dans le TEXTE : disponible même si la date est invalide
    out["annee"] = df[dob].str.extract(r"(\d{4})")[0]
    # sexe codé M/F ; manquant pour la facturation
    out["sexe"] = df[sexe].map(SEXE_MAP) if sexe else np.nan
    # libellé de service ramené à 5 valeurs
    out["service"] = df["service"].map(SERVICE_MAP)
    # vérité terrain : évaluation uniquement
    out["true_id"] = df["true_id"].values
    return out
    
# Le Dossier patient stocke « NOM Prénom » dans une seule colonne : le prénom est le dernier mot
dp_pat["nom"] = dp_pat["nom_complet"].str.rsplit(" ", n=1).str[0]
dp_pat["prenom"] = dp_pat["nom_complet"].str.rsplit(" ", n=1).str[1]

# Tableau pivot : les trois sources empilées (1 725 lignes), puis clés de comparaison
# normalisées
R = pd.concat([
    vers_pivot(adm, "adm", "adm_id", "nom", "prenom", "date_naissance", "%d/%m/%Y", sexe="sexe"),
    vers_pivot(dp_pat, "dp", "dp_id", "nom", "prenom", "date_naissance", "%Y-%m-%d", sexe="sexe"),
    vers_pivot(fac, "fac", "fac_id", "nom", "prenom", "date_naissance", "%d-%m-%Y"),
], ignore_index=True)
R["nom_n"] = R["nom"].map(normaliser)
R["prenom_n"] = R["prenom"].map(normaliser)
print(R.shape)      # attendu : (1725, 11) = 603 + 569 + 553 lignes, 11 colonnes
print(R.head(3).to_string())
```

## 2.3 Blocking

```python
# Fonction 1 : toutes_paires(indices) -> toutes les paires (a, b), a < b, d'une liste
# d'enregistrements

# Ex. : [4, 7, 9] -> {(4, 7), (4, 9), (7, 9)}
def toutes_paires(indices):
    indices = sorted(indices)
    paires = set()
    for a in range(len(indices)):
        for b in range(a + 1, len(indices)):
            paires.add((indices[a], indices[b]))
    return paires
# Blocking standard : on ne compare que les enregistrements qui partagent la même clé de blocage
# clé = année de naissance + initiale du nom (ex. « 1961_l »)
requis = ["R", "V", "total", "P_std", "P_sn", "toutes_paires",
          "paires_blocking", "paires_sorted_neighborhood"]
manquants = [nom for nom in requis if nom not in globals()]
print("Manquants :", manquants if manquants else "aucun, tu peux lancer la figure")
R["cle"] = R["annee"].fillna("") + "_" + R["nom_n"].str[:1]

def paires_blocking(cle):
    paires = set()
    # un bloc = les enregistrements de même clé
    for _, indices in R.groupby(cle).groups.items():
        # on compare tout le monde, à l'intérieur du bloc
        paires |= toutes_paires(indices)
    return paires
    
# Sorted Neighborhood : on trie les enregistrements sur une clé, puis on fait glisser une
# fenêtre.
# Fenêtre de 6 = chaque enregistrement est comparé aux 5 suivants dans l'ordre trié.
# Plus la fenêtre est large, plus on retrouve de vraies paires, mais plus on compare de paires.

# def paires_sorted_neighborhood(cle_tri, fenetre=5):
    # TODO : trier R selon `cle_tri` ; pour chaque position p de l'ordre trié, former les paires
    # avec les (fenetre - 1) enregistrements suivants ; retourner l'ensemble des paires (a, b),
    # a < b
    # raise NotImplementedError

def paires_sorted_neighborhood(cle_tri, fenetre=5):
    """
    Génère des paires d'enregistrements voisins après tri.

    cle_tri : série contenant les clés de tri
    fenetre : taille de la fenêtre glissante

    Retourne un ensemble de paires (a, b), avec a < b.
    """

    if fenetre < 1:
        raise ValueError("La fenêtre doit être supérieure ou égale à 1.")

    # Trier les enregistrements selon leur clé
    ordre = cle_tri.sort_values(kind="mergesort").index.tolist()

    paires = set()

    # Parcourir les enregistrements dans l'ordre trié
    for p in range(len(ordre)):

        # Examiner les enregistrements suivants dans la fenêtre
        for q in range(p + 1, min(p + fenetre, len(ordre))):

            a = ordre[p]
            b = ordre[q]

            # Garantir que a < b
            if a < b:
                paires.add((a, b))
            else:
                paires.add((b, a))

    return paires

# Mesures : taux de réduction et complétude des paires

vrai = R["true_id"].to_numpy()
V = set()    # vraies paires : même patient réel
for _, indices in R.groupby("true_id").groups.items():
    V |= toutes_paires(indices)
N = len(R)
total = N * (N - 1) // 2    # nombre de paires possibles sans blocking

def bilan(nom, P):
    reduction = 100 * (1 - len(P) / total)   # part de comparaisons évitées
    completude = 100 * len(P & V) / len(V)   # part des vraies paires conservées
    print(f"{nom:28s} paires={len(P):7d} réduction={reduction:5.1f} % complétude={completude:5.1f} %")
        
P_std = paires_blocking(R["cle"])
P_sn = paires_sorted_neighborhood(R["nom_n"] + R["prenom_n"] + R["annee"].fillna(""), fenetre=6)
bilan("Blocking standard", P_std)
bilan("Sorted Neighborhood (w=6)", P_sn)
bilan("Union des deux", P_std | P_sn)
```

## 2.4 Attribut et mesure de similarité

```python
from rapidfuzz.distance import JaroWinkler, Levenshtein

# ---------------------------------------------------------------------
# Mesures de similarité : toutes retournent un score entre 0 (différent) et 1 (identique)
# ---------------------------------------------------------------------

def jw(a, b):
    """Jaro-Winkler : tolère les transpositions et favorise le début commun (adapté aux noms)."""
    return JaroWinkler.normalized_similarity(a, b)
def lev(a, b):
    """Levenshtein normalisée : 1 - (nombre d'éditions minimal) / longueur maximale."""
    return Levenshtein.normalized_similarity(a, b)
def bigrammes(s):
    """Ensemble des paires de lettres consécutives : « lahlou » -> {la, ah, hl, lo, ou}."""
    return {s[i:i + 2] for i in range(len(s) - 1)} or {s}
def jaccard(a, b):
    """Jaccard sur bigrammes : |A ∩ B| / |A ∪ B|."""
    A, B = bigrammes(a), bigrammes(b)
    return len(A & B) / len(A | B)

def vecteur_bigrammes(s):
    """Vecteur de comptes : à chaque bigramme, le nombre de fois où il apparaît dans s.
    « anna » -> {"an": 1, "nn": 1, "na": 1}"""
    v = {}
    for k in range(len(s) - 1):
        g = s[k:k + 2]
        # v.get(g, 0) vaut 0 si le bigramme n'a pas encore été vu
        v[g] = v.get(g, 0) + 1
    return v
    
def cos(a, b):
    """Cosinus entre les vecteurs de bigrammes de a et de b : produit scalaire / produit des normes.
    1 = mêmes bigrammes dans les mêmes proportions ; 0 = aucun bigramme commun."""
    va, vb = vecteur_bigrammes(a), vecteur_bigrammes(b)
    produit = 0
    for g in va:
        # seuls les bigrammes communs contribuent au produit scalaire
        if g in vb:
            produit += va[g] * vb[g]
    norme_a = sum(x * x for x in va.values()) ** 0.5
    norme_b = sum(x * x for x in vb.values()) ** 0.5
    if norme_a == 0 or norme_b == 0:
        return 0.0
    return produit / (norme_a * norme_b)
    
def sim_date(a, b):
    """
    Compare deux dates de naissance.

    Retourne :
    - 1.0 si les dates sont identiques ;
    - 0.8 si une erreur plausible est détectée ;
    - 0.0 sinon.
    """

    # Vérifier si une date est manquante
    if pd.isna(a) or pd.isna(b):
        return 0.0

    # Convertir les valeurs en dates
    ta = pd.to_datetime(a, errors="coerce")
    tb = pd.to_datetime(b, errors="coerce")

    # Vérifier si les dates sont valides
    if pd.isna(ta) or pd.isna(tb):
        return 0.0

    # Cas 1 : dates identiques
    if ta == tb:
        return 1.0

    # Cas 2 : écart d'au plus un jour
    if abs((ta - tb).days) <= 1:
        return 0.8

    # Cas 3 : jour et mois inversés, avec la même année
    if (
        ta.day == tb.month
        and ta.month == tb.day
        and ta.year == tb.year
    ):
        return 0.8

    # Cas 4 : même jour et même mois, année différente d'un an
    if (
        ta.day == tb.day
        and ta.month == tb.month
        and abs(ta.year - tb.year) == 1
    ):
        return 0.8

    # Aucun cas de similarité détecté
    return 0.0

def egal(a, b):
    """Égalité exacte (1 ou 0) ; 0,5 (neutre) si l'une des deux valeurs est manquante."""
    if pd.isna(a) or pd.isna(b):
        return 0.5
    return float(a == b)
    
# Les neuf similarités qui décrivent une paire
COLONNES = ["nom_jw", "nom_lev", "nom_jac", "pre_jw", "pre_lev", "cos_np", "dob_sim", "sexe_eq", "svc_eq"]

def vecteur_similarites(i, j):
    """Vecteur des neuf similarités entre les enregistrements i et j (dans l'ordre de COLONNES)."""
    a, b = R.loc[i], R.loc[j]
    return [jw(a.nom_n, b.nom_n), lev(a.nom_n, b.nom_n), jaccard(a.nom_n, b.nom_n),
        jw(a.prenom_n, b.prenom_n), lev(a.prenom_n, b.prenom_n),
        cos(a.nom_n + " " + a.prenom_n, b.nom_n + " " + b.prenom_n),
        sim_date(a.dob, b.dob), egal(a.sexe, b.sexe), egal(a.service, b.service)]


# Paires candidates = union des deux blocking ; une ligne de X par paire
PAIRES = sorted(P_std | P_sn)
X = pd.DataFrame([vecteur_similarites(i, j) for i, j in PAIRES], columns=COLONNES)
X["i"] = [p[0] for p in PAIRES]     # indice du 1er enregistrement dans R
X["j"] = [p[1] for p in PAIRES]    # indice du 2e enregistrement dans R
X["y"] = [int(vrai[i] == vrai[j]) for i, j in PAIRES]    # étiquette : 1 = même patient
print(X.shape)   # (nombre de paires, 12) : 9 similarités + i, j, y
print("vraies paires :", X.y.sum(), f"({100 * X.y.mean():.1f} %)")
# similarité moyenne des fausses (0) et vraies (1) paires
print(X.groupby("y")[COLONNES].mean().round(2).T)
```

```python
# ---------------------------------------------------------------------
# Technique 1 : DÉTERMINISTE (règles booléennes strictes)
# ---------------------------------------------------------------------
def regles(row):
    """Applique les règles à une paire (row = une ligne de X) et retourne 0, 1 ou 2."""
    nom_ok = row.nom_jw == 1.0   # noms normalisés strictement identiques
    pre_ok = row.pre_jw == 1.0
    dob_ok = row.dob_sim == 1.0 # dates de naissance strictement identiques
    if nom_ok and pre_ok and dob_ok: # R1 : nom + prénom + naissance identiques -> Match
        return 2
    # TODO : règle R2 -> « Possible Match » (1) si la naissance est identique et si au moins un
    # des deux noms l'est
    return 0 # sinon : No Match
X["cls_regles"] = X.apply(regles, axis=1) # applique `regles` à chaque ligne de X
```

### 2.5.4 Technique probabiliste (Fellegi-Sunter)

```python
print("poids_accord :", poids_accord)
print("poids_desaccord :", poids_desaccord)
print("pi :", pi)
print("Type de G :", G.dtype)
print("Forme de G :", G.shape)
```

```python
# ---------------------------------------------------------------------
# Technique 3 : PROBABILISTE (Fellegi-Sunter, paramètres estimés par EM)
# ---------------------------------------------------------------------
def accords(df):
    """Transforme les similarités en accord (1) / désaccord (0) par attribut, selon un seuil."""
    g = pd.DataFrame({
        "nom": (df.nom_jw >= 0.92), "prenom": (df.pre_jw >= 0.92),
        "dob": (df.dob_sim >= 0.8), "sexe": (df.sexe_eq == 1.0), "service": (df.svc_eq == 1.0)})
    return g.astype(int).to_numpy()

def em_fellegi_sunter(G, n_iter=100):
    """Estime, sans étiquettes : m = P(accord | même patient), u = P(accord | patients différents),
    pi = proportion de vraies paires. Algorithme EM : on alterne deux étapes."""
    m, u, pi = np.full(G.shape[1], 0.9), np.full(G.shape[1], 0.1), 0.1
    # valeurs initiales
    for _ in range(n_iter):
        # Étape E : probabilité que chaque paire soit un match, avec les m, u actuels
        pm = np.prod(np.where(G == 1, m, 1 - m), axis=1) * pi
        pu = np.prod(np.where(G == 1, u, 1 - u), axis=1) * (1 - pi)
        g = pm / (pm + pu)
        # Étape M : nouveaux m, u, pi à partir de ces probabilités
        m = (g[:, None] * G).sum(0) / g.sum()
        u = ((1 - g)[:, None] * G).sum(0) / (1 - g).sum()
        pi = g.mean()
        m, u = np.clip(m, 1e-4, 1 - 1e-4), np.clip(u, 1e-4, 1 - 1e-4)     # évite log(0)
    return m, u, pi


G = accords(X)
m_k, u_k, pi = em_fellegi_sunter(G)
# TODO : poids d'un accord = log2(m / u) ; poids d'un désaccord = log2((1 - m) / (1 - u))
# Poids d'un accord et d'un désaccord pour chaque attribut
poids_accord = np.log2(m_k / u_k)
poids_desaccord = np.log2((1 - m_k) / (1 - u_k))
# poids total de chaque paire
w = np.where(G == 1, poids_accord, poids_desaccord).sum(1) + np.log2(pi / (1 - pi))
X["poids_fs"] = w
X["score_fs"] = 1 / (1 + 2.0 ** (-w))   # probabilité a posteriori de match
X["cls_fs"] = classer(X["score_fs"])
print(pd.DataFrame({"m": m_k, "u": u_k, "w_accord": poids_accord, "w_desaccord": poids_desaccord},
index=["nom", "prenom", "dob", "sexe", "service"]).round(3))
print("pi estimé :", round(pi, 3))
```

```python
# ---------------------------------------------------------------------
# Évaluation sur les paires de TEST, à partir de la vérité terrain y
# ---------------------------------------------------------------------

import numpy as np

def evaluer(cls, y):
    """
    cls : décisions (0 = No Match, 1 = Possible Match, 2 = Match)
    y   : vérité terrain (0 = patients différents, 1 = même patient)
    """

    cls = np.asarray(cls)
    y = np.asarray(y).astype(int)

    # Matchs automatiquement acceptés
    VP = int(((cls == 2) & (y == 1)).sum())

    # Mauvaises fusions : matchs prédits alors que les patients diffèrent
    FP = int(((cls == 2) & (y == 0)).sum())

    # Vrais matchs rejetés automatiquement
    FN = int(((cls == 0) & (y == 1)).sum())

    # Vrais non-matchs rejetés automatiquement
    VN = int(((cls == 0) & (y == 0)).sum())

    # Paires envoyées en revue manuelle
    pos_revue = int((cls == 1).sum())

    # Vrais matchs parmi les paires à revoir
    vrais_revue = int(((cls == 1) & (y == 1)).sum())

    # Précision automatique : parmi les matchs acceptés,
    # quelle proportion correspond à de vrais matchs ?
    P = VP / (VP + FP) if (VP + FP) > 0 else 0.0

    # Rappel automatique : proportion des vrais matchs
    # identifiés automatiquement parmi toutes les vraies paires
    total_vrais = int((y == 1).sum())
    Rap = VP / total_vrais if total_vrais > 0 else 0.0

    # Rappel après revue : on suppose que le Data Steward
    # identifie correctement tous les vrais matchs envoyés en revue
    Rap_revue = (
        (VP + vrais_revue) / total_vrais
        if total_vrais > 0 else 0.0
    )

    # F1-score : moyenne harmonique de la précision et du rappel
    F1 = (
        2 * P * Rap / (P + Rap)
        if (P + Rap) > 0 else 0.0
    )

    return {
        "VP": VP,
        "FP": FP,
        "FN": FN,
        "VN": VN,
        "Précision": round(P, 3),
        "Rappel": round(Rap, 3),
        "F1": round(F1, 3),
        "% revue": round(100 * pos_revue / len(y), 1) if len(y) else 0.0,
        "Rappel après revue": round(Rap_revue, 3)
    }


# ---------------------------------------------------------------------
# Comparaison des techniques sur le jeu de TEST
# ---------------------------------------------------------------------

TECHNIQUES = {
    "Déterministe": "cls_regles",
    "Floue (pondérée)": "cls_flou",
    "Probabiliste (F-S)": "cls_fs",
    "ML : logistique": "cls_lr",
    "ML : MLP": "cls_mlp"
}

test = X[X["test"]].copy()

bilan_tech = pd.DataFrame({
    nom: evaluer(
        test[col].to_numpy(),
        test["y"].to_numpy()
    )
    for nom, col in TECHNIQUES.items()
}).T

print("=== COMPARAISON DES TECHNIQUES ===")
display(bilan_tech.round(3))


# ---------------------------------------------------------------------
# Calibration : effet du seuil Match
# Le seuil bas reste à 0,55
# ---------------------------------------------------------------------

def etude_seuils(score_col, seuils=(0.70, 0.80, 0.90)):
    lignes = {}

    for s in seuils:
        decisions = classer(
            test[score_col].to_numpy(),
            SEUIL_BAS,
            s
        )

        r = evaluer(decisions, test["y"].to_numpy())

        lignes[f"≥ {s:.2f}"] = {
            k: r[k]
            for k in ("Précision", "Rappel", "% revue")
        }

    return pd.DataFrame(lignes).T


print("\n=== ÉTUDE DES SEUILS : MODÈLE FLOU ===")
display(etude_seuils("score_flou"))

print("\n=== ÉTUDE DES SEUILS : MODÈLE MLP ===")
display(etude_seuils("score_mlp"))
```

## 2.7 Des paires aux groupes

```python
# ---------------------------------------------------------------------
# Des paires aux groupes. Fermeture transitive : si A=B et B=C, alors {A, B, C} forment un groupe.
# Union-Find : structure qui fusionne des éléments reliés. parent[x] pointe vers un représentant
# du groupe.
# ---------------------------------------------------------------------
class UnionFind:
    def __init__(self, n):
        # au départ, chaque enregistrement est seul dans son groupe
        self.parent = list(range(n))

    def find(self, x):
        """Représentant du groupe de x."""
        while self.parent[x] != x:
            # raccourci : accélère les recherches suivantes
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]
        return x

    def union(self, a, b):
        """Relie les groupes de a et de b."""
        self.parent[self.find(a)] = self.find(b)


def construire_groupes(X, score_col, bas=SEUIL_BAS, haut=SEUIL_HAUT):
    """Relie les paires Match, et les Possible Match que le Data Steward valide (simulé par la
    vérité terrain)."""
    uf = UnionFind(len(R))
    # Match : liés automatiquement
    auto = X[X[score_col] >= haut]
    # Possible Match : revue humaine
    revue = X[(X[score_col] >= bas) & (X[score_col] < haut)]
    # le Steward valide les vraies paires
    valides = revue[revue.y == 1]
    for i, j in zip(auto.i, auto.j):
        uf.union(i, j)
    for i, j in zip(valides.i, valides.j):
        uf.union(i, j)
    # identifiant de groupe de chaque enregistrement
    return [uf.find(k) for k in range(len(R))], len(revue)


R["cluster_id"], n_revue = construire_groupes(X, "score_flou")

# ---- Évaluation des groupes
g = R.groupby("cluster_id")
pur = g["true_id"].nunique().eq(1)
# groupe pur : un seul patient réel
frag = R.groupby("true_id")["cluster_id"].nunique()
# nombre de groupes par patient réel
print(f"Enregistrements : {len(R)} | patients réels : {R.true_id.nunique()}")
print(f"Groupes obtenus : {g.ngroups} | groupes purs : {100 * pur.mean():.1f} % | "
      f"patients éclatés en plusieurs groupes : {(frag > 1).sum()} | paires en revue : {n_revue}")

# ---- Cohérence : un groupe est suspect si deux de ses membres sont jugés « No Match » (risque
# de chaînage)
def score_flou_paire(i, j):
    """Score flou entre deux enregistrements quelconques (même hors blocking)."""
    v = dict(zip(COLONNES, vecteur_similarites(i, j)))
    return sum(v[c] * w for c, w in POIDS_FLOU.items())


def groupes_incoherents(R, bas=SEUIL_BAS):
    """Groupes d'au moins 3 membres dont la paire la moins similaire a un score < `bas`."""
    suspects = []
    for cid, indices in R.groupby("cluster_id").groups.items():
        if len(indices) < 3:
            continue  # un groupe de 2 est une paire déjà jugée

        # 1. Score flou de toutes les paires possibles du groupe
        scores = [score_flou_paire(i, j) for i, j in toutes_paires(indices)]

        # 2. Le pire score = le maillon le plus faible du groupe
        pire = min(scores)

        # 3. Si ce maillon est sous le seuil bas, le groupe est suspect
        if pire < bas:
            suspects.append((cid, len(indices), round(pire, 3)))
    return suspects


suspects = groupes_incoherents(R)
print("groupes à revoir (chaînage) :", len(suspects), suspects[:5])

# ---- Sorties pour la Partie 3 (Merge)
R[["src", "rec_id", "cluster_id", "true_id"]].to_csv("groupes.csv", index=False)
X.assign(decision=X["cls_flou"].map({0: "No Match", 1: "Possible Match", 2: "Match"}))[
    ["i", "j", "score_flou", "decision", "y"]].to_csv("paires_scorees.csv", index=False)
```

## 3.4 Implémenter les règles de survivance

```python
# ---------------------------------------------------------------------
# Règles de survivance : chacune filtre la liste des candidats d'un attribut
# Un candidat = {"src", "valeur", "maj"} ; une règle retourne les candidats conservés
# ---------------------------------------------------------------------
def source_prioritaire(cands, ordre):
    """Garde les candidats de la source la mieux classée parmi celles qui ont une valeur."""
    for s in ordre:
        retenus = [c for c in cands if c["src"] == s]
        if retenus:
            return retenus
    return cands


def plus_recente(cands):
    datés = [c for c in cands if c["maj"] not in (None, "") and not pd.isna(c["maj"])]
    if not datés:                       # aucune date connue : on ne filtre rien
        return cands
    mmax = max(c["maj"] for c in datés)  # dates ISO : l'ordre alphabétique = l'ordre chronologique
    return [c for c in datés if c["maj"] == mmax]


def plus_complete(cands):
    """Garde la valeur la moins lacunaire (la plus longue après nettoyage)."""
    lmax = max(len(str(c["valeur"]).strip()) for c in cands)
    return [c for c in cands if len(str(c["valeur"]).strip()) == lmax]


def plus_fiable(cands, tous):
    # pour chaque candidat : combien de candidats du groupe ont la même valeur normalisée
    soutien = [sum(normaliser(t["valeur"]) == normaliser(c["valeur"]) for t in tous) for c in cands]
    meilleur = max(soutien)
    return [c for c, s in zip(cands, soutien) if s == meilleur]


def source_unique(cands):
    """Si une seule source porte l'attribut, sa valeur est reprise sans arbitrage."""
    # None = règle non applicable
    return cands if len({c["src"] for c in cands}) == 1 else None


# Hiérarchisation par attribut : liste ordonnée de (règle, paramètres)
REGLES = {
    "nom":       [("source_prioritaire", ["adm", "fac", "dp"]), ("plus_fiable", None)],
    "prenom":    [("source_prioritaire", ["adm", "fac", "dp"]), ("plus_fiable", None)],
    "dob":       [("source_prioritaire", ["adm", "dp", "fac"]), ("plus_fiable", None)],
    "sexe":      [],  # TODO : règles issues de votre tableau de décision (Étape 2)
    "telephone": [],  # TODO
    "service":   [],  # TODO
}


def appliquer(attribut, cands):
    """Applique les règles dans l'ordre jusqu'à obtenir une seule valeur.
    Retourne (valeur, règle décisive, statut) ; statut = 'A_TRANCHER' si le cas reste ambigu."""
    cands = [c for c in cands if c["valeur"] not in (None, "") and not pd.isna(c["valeur"])]
    if not cands:
        # l'attribut n'existe dans aucune source
        return None, "aucune_source", "VIDE"
    tous = list(cands)
    for regle, param in REGLES[attribut]:
        if regle == "source_prioritaire":
            nouv = source_prioritaire(cands, param)
        elif regle == "plus_recente":
            nouv = plus_recente(cands)
        elif regle == "plus_complete":
            nouv = plus_complete(cands)
        elif regle == "plus_fiable":
            nouv = plus_fiable(cands, tous)
        else:
            nouv = source_unique(cands)
        # règle non applicable : on passe à la suivante
        if nouv is None:
            continue
        cands = nouv
        # une seule valeur distincte : tranché
        if len({normaliser(c["valeur"]) for c in cands}) == 1:
            return cands[0]["valeur"], regle, "OK"
    return cands[0]["valeur"], "non_tranche", "A_TRANCHER"
    # transmis au Data Steward
```

## 3.5 Construire le Golden Record

```python
# ---------------------------------------------------------------------
# Construction du Golden Record (comparer -> appliquer la règle -> écrire la valeur)
# ---------------------------------------------------------------------
ATTRIBUTS = ["nom", "prenom", "dob", "sexe", "telephone", "service"]  # union des schémas


def construire_golden(S):
    golden, journal, xref = [], [], []
    for k, (cid, g) in enumerate(S.groupby("cluster_id"), 1):
        gid = f"GP-{k:05d}"
        rec = {"golden_id": gid, "cluster_id": cid, "nb_sources": len(g), "statut": "OK"}
        for att in ATTRIBUTS:
            cands = [{"src": r.src, "valeur": r[att], "maj": r.maj} for _, r in g.iterrows()]
            valeur, regle, statut = appliquer(att, cands)
            rec[att] = valeur
            if statut == "A_TRANCHER": rec["statut"] = "A_TRANCHER"
            ecartees = sorted({str(c["valeur"]) for c in cands
                               if c["valeur"] not in (None, "") and not pd.isna(c["valeur"])
                               and c["valeur"] != valeur})
            journal.append({"golden_id": gid, "attribut": att, "regle": regle,
                            "valeur_retenue": valeur,
                            "valeurs_ecartees": json.dumps(ecartees, ensure_ascii=False)})
        # Attributs multi-valués : on conserve plusieurs valeurs utiles au lieu d'en choisir une
        # TODO : conserver TOUS les services distincts du groupe (séparés par « | »)
        rec["services_vus"] = None
        rec["sources"] = ",".join(sorted(g["src"].unique()))
        golden.append(rec)
        xref += [{"golden_id": gid, "src": r.src, "rec_id": r.rec_id} for r in g.itertuples()]
    return pd.DataFrame(golden), pd.DataFrame(journal), pd.DataFrame(xref)


G, J, XREF = construire_golden(S)
print(G.shape, "| à trancher :", (G.statut == "A_TRANCHER").sum())
print(J.regle.value_counts().to_string())
print(G.head(3).to_string())
G.to_csv("golden_records.csv", index=False); J.to_csv("journal_fusion.csv", index=False)
XREF.to_csv("xref.csv", index=False)
```

## 3.6 Valider les Golden Records

```python
# ---------------------------------------------------------------------
# Validation du Golden Record : cohérence, complétude, exactitude
# ---------------------------------------------------------------------
from rapidfuzz.distance import JaroWinkler


def toutes_paires(valeurs):
    """Toutes les paires (a, b) de valeurs distinctes d'une liste : [1, 2, 3] -> (1, 2), (1, 3),
    (2, 3)."""
    valeurs = list(valeurs)
    paires = []
    for a in range(len(valeurs)):
        for b in range(a + 1, len(valeurs)):
            paires.append((valeurs[a], valeurs[b]))
    return paires


def dates_proches(a, b):
    """Vrai si deux dates ISO sont égales ou diffèrent d'une erreur de saisie plausible (cf.
    Partie 2)."""
    ta, tb = pd.Timestamp(a), pd.Timestamp(b)
    return (abs((ta - tb).days) <= 1
            or (ta.year == tb.year and ta.day == tb.month and ta.month == tb.day)
            or (abs(ta.year - tb.year) == 1 and (ta.month, ta.day) == (tb.month, tb.day)))


def controle_coherence(S):
    """Groupes dont deux membres se contredisent (date de naissance ou nom), signe d'une
    mauvaise fusion."""
    suspects = {}
    for cid, g in S.groupby("cluster_id"):
        motifs = set()

        # 1. Dates de naissance : on compare les valeurs distinctes deux à deux
        dobs = g["dob"].dropna().unique()
        for a, b in toutes_paires(dobs):
            if not dates_proches(a, b):
                motifs.add("dob")

        # 2. Noms : on normalise (sans accents ni casse), puis Jaro-Winkler
        noms = g["nom"].dropna().map(normaliser).unique()
        for a, b in toutes_paires(noms):
            if JaroWinkler.similarity(a, b) < 0.85:
                motifs.add("nom")

        # 3. Si au moins un motif, le groupe est suspect
        if motifs:
            suspects[cid] = sorted(motifs)
    return suspects


def controle_completude(G, obligatoires=("nom", "prenom", "dob")):
    """% de Golden Records dont tous les attributs obligatoires sont renseignés."""
    return 100 * G[list(obligatoires)].notna().all(axis=1).mean()


# valeurs réelles : ÉVALUATION UNIQUEMENT
ref = pd.read_csv("reference_patients.csv", dtype=str)


def exactitude(G, S):
    """Compare chaque Golden Record aux valeurs réelles du patient majoritaire de son groupe."""
    vrai_id = S.groupby("cluster_id")["true_id"].agg(lambda s: s.mode()[0])
    R_ = G.assign(true_id=G.cluster_id.map(vrai_id)).merge(ref, on="true_id", suffixes=("", "_ref"))
    out = {}
    for att, att_ref in [("nom", "nom"), ("prenom", "prenom"), ("dob", "date_naissance"),
                         ("sexe", "sexe"), ("telephone", "telephone"), ("service", "service")]:
        # on ne juge que les valeurs renseignées
        m = R_[att].notna()
        ref_col = att_ref + ("_ref" if att == att_ref else "")
        # valeur identique au caractère près
        strict = (R_.loc[m, att] == R_.loc[m, ref_col]).mean()
        # casse et espaces ignorés
        tolere = (R_.loc[m, att].map(normaliser) == R_.loc[m, ref_col].map(normaliser)).mean()
        out[att] = (round(100 * strict, 1), round(100 * tolere, 1))
    return pd.DataFrame(out, index=["strict (%)", "tolérante (%)"]).T


suspects = controle_coherence(S)
vrais_impurs = set(S.groupby("cluster_id").true_id.nunique().loc[lambda s: s > 1].index)
print("groupes suspects :", len(suspects), "| groupes réellement impurs :", len(vrais_impurs),
      "| détectés :", len(set(suspects) & vrais_impurs))
print("complétude (nom, prénom, naissance) :", round(controle_completude(G), 1), "%")
print(exactitude(G, S))
```

## 3.7 Réversibilité et bilan

```python

# ---------------------------------------------------------------------
# Réversibilité : défaire la fusion d'un groupe grâce à la table xref
# ---------------------------------------------------------------------

def defusionner(S, cid, affectation):
    """
    Sépare un cluster en sous-groupes selon l'étiquette réelle du patient.

    affectation : dictionnaire {rec_id: true_id}.
    Les enregistrements ayant la même étiquette restent ensemble.
    """
    S = S.copy()
    masque = S["cluster_id"] == cid

    # Séparer les enregistrements selon leur patient réel
    etiquettes = S.loc[masque, "rec_id"].map(affectation)

    # Ne modifier que les enregistrements dont l'étiquette est connue
    connus = etiquettes.notna()
    indices = etiquettes.index[connus]

    S.loc[indices, "cluster_id"] = etiquettes.loc[indices].map(
        lambda etiquette: f"{cid}-{etiquette}"
    )

    return S


# ---------------------------------------------------------------------
# Data Steward simulé : correction des groupes suspects
# ---------------------------------------------------------------------

S_corrige = S.copy()

for suspect in suspects:
    # Accepte soit un identifiant simple, soit un tuple (cid, taille, score)
    cid = suspect[0] if isinstance(suspect, tuple) else suspect

    sous = S_corrige[S_corrige["cluster_id"] == cid]

    if sous.empty:
        continue

    # Utiliser la vérité terrain pour simuler la correction manuelle
    affectation = dict(zip(sous["rec_id"], sous["true_id"]))

    S_corrige = defusionner(S_corrige, cid, affectation)


# Reconstruire les Golden Records, le journal et la table de correspondance
G2, J2, XREF2 = construire_golden(S_corrige)

# Compter les groupes qui contiennent plusieurs patients réels
imp2 = S_corrige.groupby("cluster_id")["true_id"].nunique().gt(1).sum()

print(
    f"Golden Records : {len(G)} -> {len(G2)} après défusion "
    f"| groupes impurs restants : {imp2} "
    f"| à trancher : {(G2['statut'] == 'A_TRANCHER').sum()}"
)


# ---------------------------------------------------------------------
# Bilan : effectifs, unicité et complétude
# ---------------------------------------------------------------------

print("\nEffectifs de patients :")

effectifs = {
    "Admissions": len(adm),
    "Facturation": len(fac),
    "Golden Records": len(G2),
    "Patients réels (vérité terrain)": ref["true_id"].nunique()
}

# Ajouter le dossier patient seulement si la variable dp existe
if "dp" in globals():
    effectifs["Dossier patient"] = len(dp)

print(pd.Series(effectifs).to_string())


# ---------------------------------------------------------------------
# Unicité : clé nom + prénom + date de naissance
# ---------------------------------------------------------------------

def taux_unicite(df, nom="nom", prenom="prenom", dob="dob"):
    if len(df) == 0:
        return 0.0

    cle = (
        df[nom].fillna("").map(normaliser)
        + df[prenom].fillna("").map(normaliser)
        + df[dob].fillna("").astype(str).map(normaliser)
    )

    return round(100 * cle.nunique() / len(df), 1)


print("\nUnicité (clé nom + prénom + naissance normalisés) :")

print({
    "Admissions (sources)": taux_unicite(S[S["src"] == "adm"]),
    "Golden Records": taux_unicite(G2)
})


# ---------------------------------------------------------------------
# Complétude par attribut et par source
# ---------------------------------------------------------------------

def completude_col(df, col):
    if len(df) == 0:
        return 0.0

    return round(100 * df[col].notna().mean(), 1)


sources = {
    "adm": S[S["src"] == "adm"],
    "dp": S[S["src"] == "dp"],
    "fac": S[S["src"] == "fac"],
    "golden": G2
}

# Calculer la complétude pour chaque attribut disponible
bilan = pd.DataFrame({
    source: {
        attribut: completude_col(df, attribut)
        for attribut in ATTRIBUTS
        if attribut in df.columns
    }
    for source, df in sources.items()
})

print("\nComplétude par attribut (%) :")
print(bilan.to_string())


# ---------------------------------------------------------------------
# Exactitude des Golden Records
# ---------------------------------------------------------------------

print(exactitude(G2, S_corrige))


# ---------------------------------------------------------------------
# Export des résultats
# ---------------------------------------------------------------------

G2.to_csv("golden_records.csv", index=False)
J2.to_csv("journal_fusion.csv", index=False)
XREF2.to_csv("xref.csv", index=False)

print("\nFichiers exportés :")
print("- golden_records.csv")
print("- journal_fusion.csv")
print("- xref.csv")
```

J’utilise (S.src == "adm").sum() plutôt que len(adm) pour ne dépendre que de S, ce qui évite le problème de la variable dp. La ligne verte verticale est la cible : l’idéal est que les deux barres dorées l’atteignent.

Panneau 2, la carte de chaleur. imshow colore une table (attributs en lignes, sources en colonnes) du rouge (0 %) au vert (100 %). Le Rectangle noir encadre la colonne golden pour la comparer d’un coup d’œil aux sources. Les cases vides structurelles, comme sexe pour fac, apparaissent en rouge.

Panneau 3, les haltères avant/après. On appelle exactitude deux fois : sur le Golden d’origine (G, S) et sur le corrigé (G2, S_corrige). Chaque attribut a un point rouge (avant) et un point vert (après), reliés par un trait. Plus le trait est long, plus la défusion a corrigé.

Panneau 4, l’unicité. taux_unicite mesure la part de clés distinctes (nom + prénom + naissance normalisés) parmi les lignes. Un taux proche de 100 % veut dire qu’il reste peu de doublons.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Rectangle

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

#  Panneau 1 : l'entonnoir, des sources vers les patients réels 
ax = axes[0, 0]
etapes = {
    "Admissions":            (S.src == "adm").sum(),
    "Dossier patient":       (S.src == "dp").sum(),
    "Facturation":           (S.src == "fac").sum(),
    "Golden (avant défusion)": len(G),
    "Golden (après défusion)": len(G2),
    "Patients réels":        ref.true_id.nunique(),
}
couleurs = ["lightgrey"] * 3 + ["goldenrod"] * 2 + ["seagreen"]
ax.barh(list(etapes.keys()), list(etapes.values()), color=couleurs)
for y, v in enumerate(etapes.values()):
    ax.text(v, y, f" {v}", va="center", fontweight="bold")
ax.axvline(ref.true_id.nunique(), color="seagreen", linestyle="--", linewidth=1)
ax.invert_yaxis()
ax.set_title("Des sources aux patients : la cible est la ligne verte")
ax.set_xlabel("Nombre d'enregistrements / de patients")

#  Panneau 2 : carte de chaleur de la complétude 
ax = axes[0, 1]
b = bilan.reindex(ATTRIBUTS)
im = ax.imshow(b.values, cmap="RdYlGn", vmin=0, vmax=100, aspect="auto")
ax.set_xticks(range(b.shape[1])); ax.set_xticklabels(b.columns)
ax.set_yticks(range(b.shape[0])); ax.set_yticklabels(b.index)
for r in range(b.shape[0]):
    for c in range(b.shape[1]):
        ax.text(c, r, f"{b.values[r, c]:.0f}", ha="center", va="center", fontsize=10)
# cadre autour de la colonne « golden »
ax.add_patch(Rectangle((b.shape[1] - 1.5, -0.5), 1, b.shape[0],
                       fill=False, edgecolor="black", linewidth=3))
ax.set_title("Complétude (%) : le Golden Record comble les trous des sources")
plt.colorbar(im, ax=ax, fraction=0.046)

#  Panneau 3 : exactitude avant / après défusion (haltères) 
ax = axes[1, 0]
avant = exactitude(G, S)["tolérante (%)"].reindex(ATTRIBUTS)
apres = exactitude(G2, S_corrige)["tolérante (%)"].reindex(ATTRIBUTS)
y = np.arange(len(ATTRIBUTS))
ax.hlines(y, avant, apres, color="lightgrey", linewidth=4, zorder=1)
ax.scatter(avant, y, s=120, color="indianred", label="Avant défusion", zorder=2)
ax.scatter(apres, y, s=120, color="seagreen", label="Après défusion", zorder=2)
ax.set_yticks(y); ax.set_yticklabels(ATTRIBUTS)
ax.invert_yaxis()
ax.axvline(100, color="black", linestyle=":", linewidth=1)
ax.set_xlabel("% de valeurs identiques à la vérité terrain (comparaison tolérante)")
ax.set_title("Exactitude : un trait long = la défusion a corrigé des fusions erronées")
ax.legend(loc="lower left")

#  Panneau 4 : unicité (doublons restants) 
ax = axes[1, 1]
unic = {"Admissions\n(source)": taux_unicite(S[S.src == "adm"]),
        "Golden\navant": taux_unicite(G),
        "Golden\naprès": taux_unicite(G2)}
ax.bar(unic.keys(), unic.values(), color=["lightgrey", "goldenrod", "seagreen"])
for x, v in enumerate(unic.values()):
    ax.text(x, v, f"{v} %", ha="center", va="bottom", fontweight="bold")
ax.set_ylim(0, 110)
ax.set_ylabel("% de clés (nom + prénom + naissance) uniques")
ax.set_title("Unicité : plus c'est haut, moins il reste de doublons")

imp_avant = (S.groupby("cluster_id").true_id.nunique() > 1).sum()
fig.suptitle(f"Bilan du merge : groupes impurs {imp_avant} → {imp2} après défusion",
             fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```
